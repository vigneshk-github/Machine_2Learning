```python
q1 = df['placement_exam_marks'].quantile(0.25)
q3 = df['placement_exam_marks'].quantile(0.75)
```

```python
iqr = q3 - q1
```

```python
lower_limit = q1-1.5*iqr
upper_limit = q1+1.5*iqr
```

```python
print(lower_limit,upper_limit)
```

```python
((df['placement_exam_marks'] < lower_limit) | (df['placement_exam_marks'] > upper_limit)).sum()

# the data is more then 10% so we cant trim
```
