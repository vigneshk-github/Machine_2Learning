```python
import numpy as np
import pandas as pd

# Load the dataset directly from the official CampusX GitHub repository
url = "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day26-ordinal-encoding/customer.csv"
df = pd.read_csv(url)

# Inspect random rows
print(df.sample(5))

# Keep only the ordinal columns ('review', 'education') and target ('purchased')
# Ignore 'age' (numerical) and 'gender' (nominal) for this specific tutorial
df = df.iloc[:, 2:]
print(df.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.iloc[:, 0:2], 
    df.iloc[:, -1], 
    test_size=0.2, 
    random_state=42
)

print(X_train.head())
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Define explicit order: Poor(0) < Average(1) < Good(2) and School(0) < UG(1) < PG(2)
oe = OrdinalEncoder(categories=[['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']])

# Fit only on training data, then transform both train and test sets
oe.fit(X_train)

X_train = oe.transform(X_train)
X_test = oe.transform(X_test)

print("Encoded X_train:\n", X_train[:5])
print("Learned Categories:", oe.categories_)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Fit on y_train and transform both y_train and y_test
le.fit(y_train)

print("Target Classes:", le.classes_) # ['No' -> 0, 'Yes' -> 1]

y_train = le.transform(y_train)
y_test = le.transform(y_test)

print("Encoded y_train:", y_train[:10])
```
