```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv("train.csv",usecols=['Survived','Age','Fare','SibSp','Parch'])

df.head()
```

```python
X = df.drop(columns='Survived')
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
X_train["Age_99"] = X_train["Age"].fillna(99)
X_train["Age_minus1"] = X_train["Age"].fillna(-1)

X_train["Fare_999"] = X_train["Fare"].fillna(999)
X_train["Fare_minus1"] = X_train["Fare"].fillna(-1)

# Arbitrary Value Imputation using Scikit-Learn SimpleImputer
imputer_const1 = SimpleImputer(strategy="constant", fill_value=99)
imputer_const2 = SimpleImputer(strategy="constant", fill_value=-1)

trf_arbitrary = ColumnTransformer(
    transformers=[
        ("imputer_age_99", imputer_const1, ["Age"]),
        ("imputer_fare_minus1", imputer_const2, ["Fare"]),
    ],
)

X_train_arbitrary = trf_arbitrary.fit_transform(X_train)
X_test_arbitrary = trf_arbitrary.transform(X_test)
```
