```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train1.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=["Survived"])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
X_train["Age_imputed"] = X_train["Age"]
X_test["Age_imputed"] = X_test["Age"]
```

```python
X_train.head()
```

```python
# Replacing the null values with some random values 


X_train["Age_imputed"][X_train["Age_imputed"].isnull()] = X_train["Age"].dropna().sample(X_train["Age_imputed"].isnull().sum()).values
X_test["Age_imputed"][X_test["Age_imputed"].isnull()] = X_test["Age"].dropna().sample(X_test["Age_imputed"].isnull().sum()).values
```

```python
X_train
```

```python
sns.distplot(X_train["Age"],label="Orginal",hist=False)

sns.distplot(X_train["Age_imputed"],label="Imputed",hist=False)

plt.legend()

plt.show()

```
