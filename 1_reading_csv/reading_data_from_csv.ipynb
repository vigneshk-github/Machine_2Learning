```python
# To filter out the columns we want to read from the csv file, we can use the usecols parameter in the read_csv function.
#  This allows us to specify a list of column names that we want to include in the resulting DataFrame, while ignoring the rest of the columns in the csv file.
pd.read_csv("sample_100kb_dataset.csv",usecols=['feature_1','feature_2'])
```

```python
#to skips the rows in the csv file, we can use the skiprows parameter in the read_csv function. This allows us to specify the number of rows to skip at the beginning of the file, which can be useful if the file contains metadata or other information that we don't want to include in our DataFrame.
pd.read_csv("sample_100kb_dataset.csv",skiprows=2)
```

```python
# nrows parameter in the read_csv function allows us to specify the number of rows to read from the csv file. This can be useful if we only want to read a subset of the data, or if we want to limit the amount of memory used when reading large files.
pd.read_csv("sample_100kb_dataset.csv",nrows=5)
```

```python
#Encoding Parameter in the read_csv function allows us to specify the character encoding of the csv file. This can be useful if the file contains non-ASCII characters or if we are working with files from different regions that use different encodings. By specifying the correct encoding, we can ensure that the data is read correctly and avoid issues with character representation.
# Sometime the files we get will not always be in the same encoding format, so we need to specify the encoding parameter to ensure that the data is read correctly. For example, if the file is encoded in UTF-8, we can use the following code to read it:
pd.read_csv("sample_100kb_dataset.csv",encoding='utf-8')
```

```python
#on_bad_lines lines in the csv file, we can use the on_bad_lines parameter in the read_csv function. This allows us to specify whether to skip lines that contain errors or to raise an exception when an error is encountered. By setting on_bad_lines to 'skip', we can skip any lines that contain errors and continue reading the rest of the file without interruption.
#If all rows has 8 columsn but one row has 9 columns data then it will throw an error, so we can use the on_bad_lines parameter to skip that row and read the rest of the data without any issues.
pd.read_csv("sample_100kb_dataset.csv",on_bad_lines='skip')
```

```python
#Handling dates
# If the dates are in object form we cnat apply the date functions on them, so we need to convert them to datetime format using the parse_dates parameter in the read_csv function. This allows us to specify a list of column names that we want to convert to datetime format, which can be useful for performing date-related operations and analysis on the data.
pd.read_csv("dates_dataset_100kb.csv",parse_dates=['date'])
```

```python
#Convertors 
#We can use the converters parameter in the read_csv function to apply a custom function to a specific column while reading the csv file. This allows us to transform the data in that column according to our needs, such as converting strings to lowercase, applying mathematical operations, or any other custom transformation.
# Suppose in any column we have the data in string format but we want to convert it to integer format, we can use the converters parameter to apply a custom function that converts the string to an integer. For example, if we have a column called 'feature_1' that contains string representations of integers, we can use the following code to read the csv file and convert the 'feature_1' column to integer format:
pd.read_csv("sample_100kb_dataset.csv",converters={'feature_1': lambda x: float(x)})
```
