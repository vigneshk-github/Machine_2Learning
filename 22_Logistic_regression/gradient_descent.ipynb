```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
```

```python
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=20,
)

# ==========================================
# 2. Scikit-Learn Baseline (No Regularization)
#    Note: Use penalty=None in scikit-learn >= 1.2 (was penalty='none' in older versions)
# ==========================================
lor = LogisticRegression(penalty=None, solver="sag", max_iter=5000)
lor.fit(X, y)

m_lor = -(lor.coef_[0][0] / lor.coef_[0][1])
b_lor = -(lor.intercept_[0] / lor.coef_[0][1])

x_input = np.linspace(-3, 3, 100)
y_input_lor = m_lor * x_input + b_lor
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def gd(X,y,lr=0.5,epochs=1000):
    X_aug = np.insert(X,0,1,axis=1)
    weights = np.ones(X_aug.shape[1])

    for _ in range(epochs):

        y_hat = sigmoid(np.dot(X_aug,weights))

        weights = weights + lr * (np.dot((y - y_hat),X_aug)/ X_aug.shape[0])

    return weights[1:], weights[0]


```

```python
coef_, intercept_ = gd(X, y, lr=0.5, epochs=5000)
print(f"Custom GD Coefs (w1, w2): {coef_} | Intercept (w0): {intercept_:.4f}")
print(f"Sklearn   Coefs (w1, w2): {lor.coef_[0]} | Intercept (w0): {lor.intercept_[0]:.4f}")
```

```python
m_gd = -(coef_[0] / coef_[1])
b_gd = -(intercept_ / coef_[1])
y_input_gd = m_gd * x_input + b_gd
```
