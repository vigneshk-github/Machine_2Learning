```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("titanic.csv")
```

```python
df.head()
```

```python
df["number"].unique()
```

Extracting Numbers

```python
df["numerical_number"] = pd.to_numeric(df["number"],errors='coerce',downcast='integer')
```

```python
df["numerical_number"]
```

Extracting Categorical Part

```python
df["number_categorical"] = np.where(df["numerical_number"].isnull(),df['number'],np.nan)

df.head()
```

```python
df["cabin_num"] = df['Cabin'].str.extract('(\d+)')  # captures numerical part

df["cabin_cat"] = df["Cabin"].str[0] # capture first letter

df.head()
```
