```python
import numpy as np
import pandas as pd
```

```python
date = pd.read_csv("orders.csv")
time = pd.read_csv("messages.csv")
```

```python
date.head()
```

```python
time.head()
```

First Objective would convert the date column datatype from object to data datatype

```python
date.info()
```

```python
date["date"] = pd.to_datetime(date["date"])
```

```python
date.info()
```

Extracting Year

```python
date["date_year"] = date['date'].dt.year

date.head()
```

Extracting Month

```python
date["date_month"] = date['date'].dt.month_name()

date.head()
```

Extracting Dates

```python
date["date_day"] = date['date'].dt.day

date.head()
```

Get Week

```python
date["date_dow"] = date['date'].dt.day_of_week

date.head()
```

Converting time column from object to datetime datatype

```python
time.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

Extracting hours mintues and seconds

```python
time["hour"] = time["date"].dt.hour
time["min"] = time["date"].dt.minute
time["sec"] = time["date"].dt.second

time.head()
```

Extracting time from date

```python
time["time"] = time["date"].dt.time

time.head()
```
