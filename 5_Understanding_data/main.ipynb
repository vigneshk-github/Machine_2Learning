We need to ask 7 questions to understand about the data

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
```

1. How Big is the data

```python
df.shape
```

2. How Data looks like

```python
# df.sample(5) instead of selecting starting or end rows what sample does is it randomly selects 5 rows from the dataset. This is useful for getting a quick look at the data without any bias towards the beginning or end of the dataset.

df.sample(5)
```

3. What is the data  type of cols

```python
df.info()

# Here the age column is float if we convert it to int, we dont require age in float by doing so we will reduce the size and make our algo faster. But we will lose the decimal values which is not a big deal in this case.
# Similarly we can find the columns which are not required and drop them to reduce the size of the dataset. This will help in making our algorithm faster.
```

4. Are there any missing values ?

```python
# with this we will find exactly how many null values are present in each column. This will help us in deciding whether we want to drop the column or fill the null values with some value.

df.isnull().sum()
```

5. How does the data look mathematically ?

```python
# Insights of the numerical columns in the dataset. This will help us in understanding the distribution of the data and also help us in identifying any outliers present in the data.

df.describe()
```

6. Are there any duplicate values ?

```python
df.duplicated().sum()
```

7. How is the correlation between the cols ?

```python
df.corr(numeric_only=True)
```

```python
# This will give us the correlation of all the numerical columns with respect to the target column 'Survived'. This will help us in understanding which features are more important for predicting the target variable.

# Why we need to check the correlation of the features with respect to the target variable is because it will help us in understanding which features are more important for predicting the target variable. The features which have high correlation with the target variable are more important for predicting the target variable.

# Do not drop a feature just because it is correlated with the target. High correlation with the target means the feature is actually very useful for prediction.

# Do drop features that are redundant (highly correlated with each other) to reduce noise, prevent unstable coefficients in linear models, and simplify the model


df.corr(numeric_only=True)['Survived'].sort_values(ascending=False)
```
