```python
import numpy as np
import pandas as pd

# 1. Load the 14-row Play Tennis dataset
data = pd.read_csv("play_tennis.csv")

# 2. TRAINING STAGE: Compute class priors
print(data["play"].value_counts())
py = 9 / 14   # P(Yes)
pn = 5 / 14   # P(No)

# 3. TRAINING STAGE: Build lookup probabilities via pd.crosstab()
# --- Outlook ---
print(pd.crosstab(data["outlook"], data["play"]))
pon, prn, psn = 0 / 5, 2 / 5, 3 / 5   # Overcast, Rain, Sunny | No
poy, pry, psy = 4 / 9, 3 / 9, 2 / 9   # Overcast, Rain, Sunny | Yes

# --- Temperature ---
print(pd.crosstab(data["temp"], data["play"]))
pcool_no, phot_no, pmild_no = 1 / 5, 2 / 5, 2 / 5
pcool_yes, phot_yes, pmild_yes = 3 / 9, 2 / 9, 4 / 9

# --- Humidity ---
print(pd.crosstab(data["humidity"], data["play"]))
phigh_no, pnormal_no = 4 / 5, 1 / 5
phigh_yes, pnormal_yes = 3 / 9, 6 / 9

# --- Wind ---
print(pd.crosstab(data["wind"], data["play"]))
pstrong_no, pweak_no = 3 / 5, 2 / 5
pstrong_yes, pweak_yes = 3 / 9, 6 / 9

# 4. TESTING STAGE: Query = (Outlook=Sunny, Temp=Hot, Humidity=High, Wind=Weak)
pyes = py * psy * phot_yes * phigh_yes * pweak_yes
pno  = pn * psn * phot_no  * phigh_no  * pweak_no

print(f"Score(Yes): {pyes:.6f}")   # 0.007055
print(f"Score(No):  {pno:.6f}")    # 0.027429
print("Prediction:", "Yes" if pyes > pno else "No")  # Output: No
```
