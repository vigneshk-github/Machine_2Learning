```python
# We can check the skewness of the Age column to understand the distribution better. A skewness value greater than 0 indicates a right-skewed distribution, while a value less than 0 indicates a left-skewed distribution. A value close to 0 suggests a symmetric distribution.

# Skew means the degree of asymmetry in the distribution. Like a right-skewed distribution has a long tail on the right side. Example: Income distribution is right-skewed because most people earn less, and a few people earn a lot more.

# If the data has more skewness the it means that the data is not normally distributed. In such cases, we can apply transformations like log transformation, square root transformation, etc., to make the data more normal.

# If we wont takle the skewness then it can lead to problems in the model performance. For example, if we have a right-skewed distribution and we use a linear regression model, the model will be biased towards the majority of the data points and will not perform well on the minority of the data points.


df["Age"].skew()
```
